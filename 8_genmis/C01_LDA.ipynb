```python
def get_stop_dict(file):
    content = open(file,encoding="utf-8")
    word_list = []
    for c in content:
        c = re.sub('\n|\r','',c)
        word_list.append(c)
    return word_list
```

```python
import jieba
## 使用自定义词典
dict = 'C01_LDA/stop_dic/dict.txt'
stopword_dict = 'C01_LDA/stop_dic/stopwords.txt'
stopwords = get_stop_dict(stopword_dict)
jieba.load_userdict(dict)
## 加载停用词表
# stoplist = list(pd.read_csv('stopword.txt',names = ['w'], sep = '\t',encoding='UTF-8').w)
```
