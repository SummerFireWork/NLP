# 模型生成函数

```python
def get_lda_model(num_topics, corpus, dictionary,  text):
    x = [] # x轴
    perplexity_values = [] # 困惑度
    coherence_values = [] # 一致性
    
    for i in range(1, num_topics+1):
        lda_model = gensim.models.ldamodel.LdaModel(corpus=corpus, id2word=dictionary, num_topics=i, passes=10, chunksize=2000, iterations=400)
        perplexity = lda_model.log_perplexity(corpus)
        coherence_model_lda = gensim.models.CoherenceModel(model=lda_model, texts=text, dictionary=dictionary, coherence='c_v')
        coherence_lda = coherence_model_lda.get_coherence()
        
        x.append(i)
        perplexity_values.append(perplexity)
        coherence_values.append(coherence_lda)
        
    return x, perplexity_values, coherence_values
```

## 读取停用词

```python
stop_words = []
with open('stop_dic/stopwords.txt', 'r', encoding='utf-8') as f:
    for line in f.readlines():
        stop_words.append(line.strip())
```

## 读取语料库

```python
f = open('data/data.txt', 'r', encoding='utf-8')
sentence = f.read()
# text = re.sub(r'[^\u4e00-\u9fa5]', "", text)
# text = re.sub(r'\s+', "", text)
# text = re.sub(r'\n+', "", text)
# text = re.sub(r'\u3000+', "", text)

import jieba

text = []

for word in jieba.lcut(sentence):
    if word not in stop_words:
        text.append(word)
```

## 生成词袋模型所需数据

```python
import gensim.corpora as corpora

dictionary = gensim.corpora.Dictionary([text])
corpus = [dictionary.doc2bow(tmp) for tmp in text]
```

## 主题建模

```python
num_topics = 10
x, perplexity_values, coherence_values = get_lda_model(num_topics, corpus, dictionary, text)
```
